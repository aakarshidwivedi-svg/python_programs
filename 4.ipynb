```python
tup = (1, 2, 3, 4, 5)
tup2 = (4, 5, 6, 7, 8)


common = []
for item in tup:
    if item in tup2:
        common.append(item)

print("Tuple 1:", tup)
print("Tuple 2:", tup2)
print("Common elements:", tuple(common))
```

```python
my_list = [1, 2, 3, 4]


my_tuple = tuple(my_list)

print("List  :", my_list)
print("Tuple :", my_tuple)
```

```python
my_tuple = (5, 6, 7, 8)


my_list = list(my_tuple)

print("Tuple :", my_tuple)
print("List  :", my_list)
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print("Set 1:", set1)
print("Set 2:", set2)
print("Union (all elements)     :", set1 | set2)
print("Intersection (common)    :", set1 & set2)
print("Difference (set1 - set2) :", set1 - set2)
```

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

unique_numbers = list(set(numbers))

print("With duplicates   :", numbers)
print("Without duplicates:", unique_numbers)
```
