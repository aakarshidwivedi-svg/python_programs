```python

numbers = [45, 12, 78, 3, 56]
small = numbers[0]
for num in numbers:
    if num < small:
        small = num

print("The list", numbers)
print("Smallest number", small)
```

```python
numbers = [10, 20, 30, 40, 50]

total = 0
for num in numbers:
    total = total + num


average = total / len(numbers)

print("Numbers", numbers)
print("Total =", total)
print("Average =", average)
```

```python
numbers = [1, 2, 3, 2, 4, 2, 5]
target = 2        

count = 0
for num in numbers:
    if num == target:
        count = count + 1

print("List:", numbers)
print(target, "occurs", count, "times")
```

```python
numbers = [1, 2, 3, 4, 5]

reversed_list = numbers[::-1]

print("Original list:", numbers)
print("Reversed list:", reversed_list)
```

```python
numbers = [45, 12, 78, 3, 56]
print("Before sorting:", numbers)


numbers.sort()

print("After sorting",numbers)
```

```python
tup1 = (1, 2, 3, 4, 5)
tup2 = (4, 5, 6, 7, 8)

common = []
for item in tup1:
    if item in tup2:
        common.append(item)

print("Tuple 1:", tup1)
print("Tuple 2:", tup2)
print("Common elements:", tuple(common))
```

```python
my_list =[1, 2, 3, 4]


my_tuple = tuple(my_list)

print("List ", my_list)
print("Tuple:", my_tuple)
```

```python
my_tuple = (5, 6, 7, 8)

my_list = list(my_tuple)

print("Tuple :", my_tuple)
print("List:", my_list)
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print("Set 1", set1)
print("Set 2", set2)
print("Union    ", set1 | set2)
print("Intersection ", set1 & set2)
print("Difference", set1 - set2)
```

```python
numbers = [1, 2, 2, 3, 4, 4, 5]


unique_numbers = list(set(numbers))

print("With duplicates ", numbers)
print("Without duplicates", unique_numbers)
```
